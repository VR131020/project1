```python
def retrieve(query, top_k=3):

    query_embedding = model.encode(
        [query],
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    scores = cosine_similarity(
        query_embedding,
        document_embeddings
    )[0]

    top_indices = np.argsort(scores)[::-1][:top_k]

    results = []

    for idx in top_indices:
        results.append({
            "rank": len(results) + 1,
            "score": float(scores[idx]),
            "sanskrit": documents[idx],
            "english": english_documents[idx]
        })

    return results
```

```python
def recall_at_k(query_df, documents, k=3):

    correct = 0

    for _, row in query_df.iterrows():

        query_embedding = model.encode(
            [row["query"]],
            normalize_embeddings=True
        )

        doc_embeddings = model.encode(
            documents,
            normalize_embeddings=True
        )

        scores = cosine_similarity(
            query_embedding,
            doc_embeddings
        )[0]

        top_indices = np.argsort(scores)[::-1][:k]

        retrieved_docs = [
            documents[i]
            for i in top_indices
        ]

        if row["positive"] in retrieved_docs:
            correct += 1

    return correct / len(query_df)
```

```python
print("Recall@1:", recall_at_k(query_df, documents, k=1))
print("Recall@3:", recall_at_k(query_df, documents, k=3))
print("Recall@5:", recall_at_k(query_df, documents, k=5))
```

```python
def calculate_mrr(query_df, documents):

    reciprocal_ranks = []

    doc_embeddings = model.encode(
        documents,
        normalize_embeddings=True
    )

    for _, row in query_df.iterrows():

        query_embedding = model.encode(
            [row["query"]],
            normalize_embeddings=True
        )

        scores = cosine_similarity(
            query_embedding,
            doc_embeddings
        )[0]

        ranked_indices = np.argsort(scores)[::-1]

        rank = None

        for position, idx in enumerate(ranked_indices, start=1):

            if documents[idx] == row["positive"]:
                rank = position
                break

        if rank is not None:
            reciprocal_ranks.append(1 / rank)
        else:
            reciprocal_ranks.append(0)

    return np.mean(reciprocal_ranks)
```

```python
mrr = calculate_mrr(query_df, documents)

print("MRR:", mrr)
```

```python
def calculate_mrr(model, documents, query_df):
    reciprocal_ranks = []

    doc_embeddings = model.encode(
        documents,
        normalize_embeddings=True,
        convert_to_numpy=True
    )

    for _, row in query_df.iterrows():

        query_embedding = model.encode(
            [row["query"]],
            normalize_embeddings=True,
            convert_to_numpy=True
        )

        scores = cosine_similarity(
            query_embedding,
            doc_embeddings
        )[0]

        ranked_indices = np.argsort(scores)[::-1]

        positive_index = documents.index(row["positive"])

        rank = np.where(ranked_indices == positive_index)[0][0] + 1

        reciprocal_ranks.append(1 / rank)

    return np.mean(reciprocal_ranks)
```

```python
baseline_mrr = calculate_mrr(
    SentenceTransformer("intfloat/multilingual-e5-small"),
    documents,
    query_df
)

finetuned_mrr = calculate_mrr(
    model,
    documents,
    query_df
)

print("Baseline MRR:", baseline_mrr)
print("Fine-tuned MRR:", finetuned_mrr)
```
